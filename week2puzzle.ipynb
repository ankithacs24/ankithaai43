```python



def print_board(state):
    """Helper function to print the 3x3 puzzle board."""
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()

def dfs_8_puzzle(initial_state, goal_state=(1, 2, 3, 4, 5, 6, 7, 8, 0), max_depth=20):
    """
    Solves the 8-puzzle using iterative Depth-First Search with a depth limit.
    State is represented as a tuple of 9 integers (0-8), where 0 is the empty tile.
    """
    # Stack stores tuples of: (current_state, path_to_this_state, current_depth)
    stack = [(initial_state, [initial_state], 0)]
    visited = set()

    while stack:
        current_state, path, depth = stack.pop()

        # Check if we reached the goal
        if current_state == goal_state:
            return path

        # If max depth is reached, do not expand further along this branch
        if depth >= max_depth:
            continue

        visited.add(current_state)

        # Find index of the blank space (0)
        idx = current_state.index(0)
        row, col = divmod(idx, 3)

        # Possible movements (Row delta, Column delta): Up, Down, Left, Right
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        # Generate valid successor states
        for dr, dc in moves:
            nr, nc = row + dr, col + dc
            if 0 <= nr < 3 and 0 <= nc < 3:
                new_idx = nr * 3 + nc

                # Swap 0 with the neighbor tile
                temp_list = list(current_state)
                temp_list[idx], temp_list[new_idx] = temp_list[new_idx], temp_list[idx]
                next_state = tuple(temp_list)

                # Push to stack if not visited or already in the current path
                if next_state not in visited and next_state not in path:
                    stack.append((next_state, path + [next_state], depth + 1))

    return None

# --- Example Usage ---
if __name__ == "__main__":
    # Define an initial state that is close to the goal (to fit within a small depth limit)
    # Goal state is (1, 2, 3, 4, 5, 6, 7, 8, 0)
    start_state = (1, 2, 3, 0,4, 6, 7, 5, 8)  # Just a few moves away from the goal

    print("Initial Board:")
    print_board(start_state)

    solution_path = dfs_8_puzzle(start_state, max_depth=15)

    if solution_path:
        print(f"Solution found in {len(solution_path) - 1} steps!\n")
        for step_num, state in enumerate(solution_path):
            print(f"Step {step_num}:")
            print_board(state)
    else:
        print("No solution found within the specified depth limit.")


```
